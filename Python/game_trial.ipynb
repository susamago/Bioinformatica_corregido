```python
import turtle

screen = turtle.Screen()
screen.title("My game")

player = turtle.Turtle()
player.shape("turtle")
player.penup()

def move_left():
    player.setx(player.xcor() - 20)

def move_right():
    player.setx(player.xcor() + 20)

def move_up():
    player.setx(player.xcor() - 20)

def move_down():
    player.setx(player.xcor() + 20)

screen.listen()
screen.onkey(move_left, "Left")
screen.onkey(move_right, "Right")
screen.onkey(move_up, "Up")
screen.onkey(move_down, "Down")

turtle.done()
```
